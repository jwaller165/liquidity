```sql
WITH params AS (SELECT DATE'2026-10-31' AS report_date),		
pg91_book AS (		
  SELECT trim(Cusip) AS Cusip, As_Of_Date, CustomerAccount, PledgeName,		
         SUM(PledgedBookValue) AS PledgedBookValue		
  FROM silver_dev.tpg_reports.pg91_pledges_to_pledgee_portfolio_consolidate		
  GROUP BY 1,2,3,4		
),		
unpledged AS (SELECT    u.as_of_date             		
AS AsOfDate,		
    trim(u.Cusip)            AS Cusip,		
    u.SecurityDesc           AS CusipDescription,		
    u.CustomerName           AS CustomerAccount,		
    pr.CustomerAccountNumber AS Portfolio,		
    'Unpledged'              AS PledgeStatus,		
    ''                       AS PledgedToInstitution,		
    u.SafeKeepingLocation    AS SafekeepingLocation,		
    u.StatedMaturityDate     AS MaturityDate,		
    u.CpnRate                AS CouponRate,		
    u.UnpledgedAmount        AS OriginalPar,		
    u.UnpledgedParValue      AS ParValue,		
    u.UnpledgedBookValue     AS BookValue,		
    u.UnpledgedMarketValue   AS FairValue,		
    CAST(0 AS DECIMAL(24,2))                        AS PledgedMarketValue,		
    CAST(COALESCE(u.UnpledgedMarketValue, 0) AS DECIMAL(24,2)) AS UnpledgedMarketValue		
  FROM silver_dev.tpg_reports.pg92_unpledged_securities u		
  JOIN params p ON u.as_of_date = p.report_date		
  JOIN silver_dev.tpg_reports.position_report pr		
    ON u.TransNo = pr.PositionID AND u.as_of_date = pr.AsOfDate		
  WHERE u.HoldingCompName = 'PINNACLE FINANCIAL PARTNERS INC'),		
pledged AS (SELECT g.AsOfDate, trim(g.Cusip) AS Cusip,		
    g.CusipDescription,		
    g.CustomerAccount,		
    g.CustomerAccount          AS Portfolio,		
    'Pledged'                  AS PledgeStatus,		
    g.PledgedToInstitutionName AS PledgedToInstitution,		
    g.SafekeepingLocationName  AS SafekeepingLocation,		
    g.MaturityDate,		
    g.CouponRate,		
    g.PledgedOriginalPar       AS OriginalPar,		
    g.PledgeParValue           AS ParValue,		
    b.PledgedBookValue         AS BookValue,		
    g.PledgeFairValue          AS FairValue,		
    CAST(COALESCE(g.PledgeFairValue, 0) AS DECIMAL(24,2)) AS PledgedMarketValue,		
    CAST(0 AS DECIMAL(24,2))                        AS UnpledgedMarketValue		
  FROM silver_dev.tpg_reports.pledgee g		
  JOIN params p ON g.AsOfDate = p.report_date		
  JOIN pg91_book b		
    ON trim(g.Cusip) = b.Cusip		
    AND g.AsOfDate = b.As_Of_Date		
    AND g.CustomerAccount = b.CustomerAccount		
    AND g.PledgedToInstitutionName = b.PledgeName)		
SELECT * FROM pledged		
UNION ALL		
SELECT * FROM unpledged		
ORDER BY Cusip, PledgeStatus, PledgedToInstitution		
```

```sql
CREATE OR REPLACE TEMP VIEW gl_lkup AS
SELECT
  dim_dates.last_business_day_of_month as asofdate,
  regexp_replace(tm1.projects, 'No Project', '') as legacy_bank,
  dim_dates.date as eomdate,
  gl.`Level 4 ID` as level4_id,
  gl.`Level 5 ID` as level5_id,
  gl.`Account Desc` as gl_bs_acct_name,
  gl.`Account ID` as gl_bs_acct,
  SUM(CAST(tm1.vvalue AS DOUBLE)) as gl_balance
--   tm1.Entities as entity,
-- e.entity_desc,
--   e.Parent1,
--   e.LOB_LVL05_node,
--     e.LOB_LVL05_description
-- gl.`Z_Account Attributes.ReconType` as Recon_Type,
-- gl.`Z_Account Attributes.ReconGroup` as Recon_Group
FROM
  silver_prod.p021.tm1_financetrialbalancepbi tm1
    JOIN silver_prod.reference_data.d_gl_account gl
      ON tm1.Accounts = gl.`Account ID`
    JOIN finance.default.dim_dates dim_dates
      ON dim_dates.date = TO_DATE(tm1.prcs_dte, 'yyyyMMdd')
    LEFT JOIN silver_prod.reference_data.d_entity_hierarchy_wide e
      ON e.entity_id = tm1.Entities
WHERE
  Versions = 'ACTUAL'
  AND DataSources = 'TotalCur'
  AND FinanceMeasures = 'EndBal'
  -- Deposits plus Repos (2328128 sits under Level 4 FUNDREP / Level 5 REPUR, not DEPOSITS)
  AND (
    gl.`Level 4 ID` = 'DEPOSITS'
    OR gl.`Level 5 ID` = 'REPUR'
    OR gl.`Level 6 ID` IN ('FUNDSPUR', 'NOTESPAYREG', '2411007', '2410009')
  )
  AND RIGHT(Periods, 3) = 'YTD'
  AND dim_dates.date = '2026-10-31'
  AND dim_dates.date = dim_dates.end_of_month
GROUP BY
  ALL;

SELECT
  *
FROM
  gl_lkup
```
